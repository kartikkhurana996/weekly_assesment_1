```python
armstrong = []
for num in range(1, 10000):
    digits = len(str(num))
    total = 0
    temp = num
    while temp > 0:
        digit = temp % 10
        total += digit ** digits        #formula of calculating armstrong value
        temp //= 10
    if total == num:
        armstrong.append(num)
print(", ".join(map(str, armstrong)))
print("Count:", len(armstrong))
```

ANS.33

```python
n = 20

def is_prime(num):
    if num < 2:
        return False
    i = 2
    while i * i <= num:
        if num % i == 0:          #prime numbers : numbers that are divisible by themselves
            return False
        i += 1
    return True

print("Prime" if is_prime(n) else "Not prime")
prime_sum = sum(num for num in range(2, n + 1) if is_prime(num))
print("Sum of primes:", prime_sum)
```

```python
sentence = "Hello Python 123!"
total_characters = len(sentence)      #cheching total characters
words = vowels = consonants = digits = spaces = special = 0
in_word = False

for ch in sentence:
    if ch.isalpha():
        if ch.lower() in "aeiou":       #present in sentence
            vowels += 1
        else:
            consonants += 1
        if not in_word:
            words += 1
            in_word = True
    elif ch.isdigit():
        digits += 1
        in_word = False
    elif ch.isspace():
        spaces += 1
        in_word = False
    else:
        special += 1            #cheching special characters
        in_word = False

print("Total characters:", total_characters)
print("Total words:", words)
print("Vowels:", vowels)
print("Consonants:", consonants)
print("Digits:", digits)
print("Spaces:", spaces)
print("Special characters:", special)
```

```python
prices = (3845, 3902, 3878, 3955, 4012, 3998, 4055)
days = ("Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun")
maximum = max(prices)
minimum = min(prices)
average = sum(prices) / len(prices)
max_day = days[prices.index(maximum)]
min_day = days[prices.index(minimum)]

biggest_gain = prices[1] - prices[0]      #gain = selling price - cost price
gain_day = 1
for i in range(2, len(prices)):
    gain = prices[i] - prices[i - 1]
    if gain > biggest_gain:
        biggest_gain = gain
        gain_day = i

volatility = sum(abs(prices[i] - prices[i - 1]) for i in range(1, len(prices)))
print("Maximum:", maximum, max_day)
print("Minimum:", minimum, min_day)
print("Average:", round(average, 2))
print("Biggest gain:", biggest_gain)
print("Between:", days[gain_day - 1], "and", days[gain_day])
print("Total volatility:", volatility)
```

```python
def calculate_gst(amount, gst_rate=18, gst_inclusive=False):
    if not gst_inclusive:
        base_amount = amount
        gst_amount = amount * gst_rate / 100
        total_amount = amount + gst_amount
    else:                                           #basic gst calculator
        total_amount = amount
        base_amount = amount / (1 + gst_rate / 100)
        gst_amount = total_amount - base_amount
    return round(base_amount, 2), round(gst_amount, 2), round(total_amount, 2)

print(calculate_gst(100))
print(calculate_gst(118, 18, True))
```

```python
def power(base, exp):
    if exp == 0:
        return 1
    if exp < 0:                         # If the exponent is negative,
                                        # calculate the positive power first
                                        # and then take its reciprocal
        return 1 / power(base, -exp)
    return base * power(base, exp - 1)

print(power(2, 10))
print(power(5, 0))
print(power(3, -2))
```

```python
import numpy as np
sales = np.array([12000, 18500, 14200, 22100, 19800, 35200, 28400])
days = np.array(["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"])

total = np.sum(sales)
average = np.mean(sales)
max_index = np.argmax(sales)                #basic numpy functions
max_day = days[max_index]
above_average = np.sum(sales > average)
growth = ((sales[-1] - sales[0]) / sales[0]) * 100
day_changes = ((sales[1:] - sales[:-1]) / sales[:-1]) * 100

print("Total weekly sales:", total)
print("Daily average:", round(average, 2))
print("Maximum sales day:", max_day)
print("Days above average:", above_average)
print("Growth Monday to Sunday:", round(growth, 2), "%")
print("Day-on-day percentage changes:", np.round(day_changes, 2))
```
